```python
import pandas as pd
import matplotlib.pyplot as plt
import json
import io
import base64
from IPython.display import HTML, display
import google.colab # Import google.colab

def generate_productivity_chart(json_data):
    # Parse the JSON string back into a Python list of dictionaries
    logged_data = json.loads(json_data)

    if not logged_data:
        print("No data to plot.")
        return

    # Convert to DataFrame
    df = pd.DataFrame(logged_data)

    # Convert 'date' to datetime objects for proper plotting
    # The format is DD/MM/YYYY, so we specify it
    df['date'] = pd.to_datetime(df['date'], format='%d/%m/%Y')

    # Aggregate productivity and damage by date
    daily_summary = df.groupby('date').agg(
        total_productivity=('productivity_8hr', 'sum'),
        total_damage_pct=('damage_pct', 'mean') # Using mean for damage percentage per day
    ).reset_index()

    # Sort by date to ensure the line chart is correct
    daily_summary = daily_summary.sort_values(by='date')

    # Create the plots for productivity and damage percentage
    fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(12, 8), sharex=True)

    # Plot 1: Daily Productivity Trend
    axes[0].plot(daily_summary['date'], daily_summary['total_productivity'], marker='o', linestyle='-', color='skyblue')
    axes[0].set_title('Daily Productivity Trend (Pcs/8hr Day)')
    axes[0].set_ylabel('Total Productivity (Pcs/8hr Day)')
    axes[0].grid(True, linestyle='--', alpha=0.6)

    # Plot 2: Patti Damage % Trend
    axes[1].plot(daily_summary['date'], daily_summary['total_damage_pct'], marker='o', linestyle='-', color='salmon')
    axes[1].set_title('Daily Patti Damage % Trend')
    axes[1].set_xlabel('Date')
    axes[1].set_ylabel('Avg. Damage %')
    axes[1].grid(True, linestyle='--', alpha=0.6)

    plt.xticks(rotation=45)
    plt.tight_layout()

    # Convert plot to base64 image
    buf = io.BytesIO()
    plt.savefig(buf, format='png')
    plt.close(fig)
    image_base64 = base64.b64encode(buf.getvalue()).decode('utf-8')

    # Calculate average productivity by product type
    avg_productivity_by_product = df.groupby('product')['productivity_8hr'].mean().reset_index()
    avg_productivity_by_product.rename(columns={'productivity_8hr': 'Average Productivity (Pcs/8hr Day)'}, inplace=True)

    # Create an HTML table for the average productivity by product
    table_html = "<h3>Average Productivity by Product Type</h3>"
    table_html += "<table border='1' style='width: 50%; border-collapse: collapse; margin-top: 20px;'>"
    table_html += "<tr><th style='padding: 8px; text-align: left;'>Product</th><th style='padding: 8px; text-align: left;'>Avg. Productivity (Pcs/8hr Day)</th></tr>"
    for index, row in avg_productivity_by_product.iterrows():
        table_html += f"<tr><td style='padding: 8px;'>{row['product']}</td><td style='padding: 8px;'>{row['Average Productivity (Pcs/8hr Day)']:.2f}</td></tr>"
    table_html += "</table>"

    # Display the image and the table in the notebook
    display(HTML(f'<img src="data:image/png;base64,{image_base64}" alt="Productivity and Damage Chart">{table_html}'))
```
